```python
edad = int(input("Tu edad cuál es? "))

if edad >= 66:
    print("Eres un adulto mayor")
elif edad >= 31 and edad <= 65:
    print("Eres un adulto")
elif edad >= 19 and edad <= 30:
    print("Eres un joven adulto")
elif edad >= 11 and edad <= 18:
    print("Eres un adolescente")
elif edad >= 1 and edad <= 10:
    print("Eres un niño")
else:
    print("Edad no válida o eres un bebé")
```
