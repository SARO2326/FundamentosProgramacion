```python
tabla = int(input("Dame la tabla: "))
inicio = int(input("Desde que numero: "))
fin = int(input("Hasta que numero: "))
paso = int(input("De cuanto en cuanto: "))

i = inicio
while i <= fin:
    print(tabla, "*", i, "=", tabla*i)
    i = i + paso
```
