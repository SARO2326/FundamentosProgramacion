```python
edad=int(input("¿Tu edad cuál es? "))
if edad >= 65:
  print("Eres un adulto")
elif edad >= 30:
  print("Eres un joven adulto")
elif edad >= 18:
  print("Eres un adolescente")
elif edad >= 10:
  print("Eres un niño")
```

```python
numero = 1
while numero <= 5:
    print(numero)
    numero = numero + 1
```

```python
tabla = int(input("Dame la tabla: "))
inicio = int(input("Desde que numero: "))
fin = int(input("Hasta que numero: "))
paso = int(input("De cuanto en cuanto: "))
while inicio >= fin:
  print("El limite final debe ser mayor que el inicial")
  fin= int(input("Dame el limite final :"))

for i in range(inicio, fin + 1, paso):
    print(tabla, "*", i, "=", tabla * i)
```
