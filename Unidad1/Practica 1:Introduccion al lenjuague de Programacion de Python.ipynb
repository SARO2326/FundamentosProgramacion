```python
tabla = int(input("Dame la tabla: "))
inicio = int(input("Desde que numero: "))
fin = int(input("Hasta que numero: "))
paso = int(input("De cuanto en cuanto: "))
while inicio >= fin:
  print("El limite final debe ser mayor que el inicial")
  fin= int(input("Dame el limite final :"))

for i in range(inicio, fin + 1, paso):
    print(tabla, "*", i, "=", tabla * i)
```

Listas en Python

```python
alumnos = ["Ana", "Luis", "Carlos"]
#podemos sobrescribir un elemnto de la lista y al hacerlo se remplaza el valor que sobrescribimos
alumnos[1]="Juan"
print(alumnos)
```

```python
alumnos = ["Ana", "Juan","Luis", "Carlos"]
#El metodo append agrega un elemento al final de la lista
alumnos.append("Sandra")
print(alumnos)
```

```python
alumnos = ["Ana", "Juan","Luis", "Carlos"]
print(alumnos)
alumnos.insert(2,"Sofia")
print(alumnos)
```

```python
alumnos = ["Ana", "Juan","Luis","Sofia","Carlos"]
print(alumnos)
#Borra un elemento de la lista especificando el valor;si existen valores repetidos ecuentra el primer valor que encuentra
alumnos.remove("Sofia")
print(alumnos)
```
